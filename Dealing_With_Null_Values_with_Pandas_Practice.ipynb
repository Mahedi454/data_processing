# Import Libraries

````python
import numpy as np
import pandas as pd
````

 Python list <b>headers</b> containing name of headers


````python
# These are the headers which you need to link to your dataset.
headers = ["symboling","normalized-losses","make","fuel-type","aspiration", "num-of-doors","body-style",
         "drive-wheels","engine-location","wheel-base", "length","width","height","curb-weight","engine-type",
         "num-of-cylinders", "engine-size","fuel-system","bore","stroke","compression-ratio","horsepower",
         "peak-rpm","city-mpg","highway-mpg","price"]
````

# Import Datasets

````python
df = pd.read_csv('car_dataset.data', names = headers)
````

````python
# To see what the data set looks like, we'll use the head() method.
df.head(15)
````

````python

df.replace("?", np.nan, inplace= True)
# replace "?" to NaN
df.head()
````

````python
df.isnull()
````

<h2>Count missing values in each column</h2>



````python
missing_data = df.isnull()
for column in headers:
  print(column)
  print(missing_data[column].value_counts())
  print(" ")
````

<h2 id="deal_missing_values">Deal with missing data</h2>



<h3>Calculate the average of the normalized-losses column </h3>


````python
avg_nor_loss = df["normalized-losses"].astype("float").mean()
print("The avg of normalized-losses is: ",avg_nor_loss)
````

<h3>Replace "NaN" by mean value in "normalized-losses" column</h3>


````python
df.replace({"normalized-losses": np.nan}, avg_nor_loss, inplace=True)
df.head(10)
````

<h3>Calculate the mean value for 'bore' column</h3>


````python
avg_bore = df["bore"].astype("float").mean()
print("The avg of bore is: ", avg_bore)
````

<h3>Replace NaN by mean value for bore column</h3>


````python
df.replace({"bore": np.nan}, avg_bore, inplace=True)
df.head(10)
````
